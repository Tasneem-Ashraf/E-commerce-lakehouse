✅ Duplicate Checks

```sql
SELECT
    customer_id,
    COUNT(*) AS row_count
FROM gold.customer_performance
GROUP BY customer_id
HAVING COUNT(*) > 1;
```

```sql
SELECT
    product_id,
    COUNT(*) AS row_count
FROM gold.product_performance
GROUP BY product_id
HAVING COUNT(*) > 1;
```

```sql
SELECT
    product_category_name,
    COUNT(*) AS row_count
FROM gold.category_performance
GROUP BY product_category_name
HAVING COUNT(*) > 1;
```

```sql
SELECT
    payment_type,
    COUNT(*) AS row_count
FROM gold.payment_performance
GROUP BY payment_type
HAVING COUNT(*) > 1;
```

✅ NULL Checks

```sql
SELECT
    COUNT(*) AS total_rows,
    COUNT(customer_id) AS customer_id_not_null,
    COUNT(total_orders) AS total_orders_not_null,
    COUNT(total_spent) AS total_spent_not_null,
    COUNT(average_order_value) AS average_order_value_not_null
FROM gold.customer_performance;
```

```sql
SELECT
    COUNT(*) AS total_rows,
    COUNT(product_id) AS product_id_not_null,
    COUNT(product_category_name) AS category_not_null,
    COUNT(total_orders) AS total_orders_not_null,
    COUNT(product_revenue) AS product_revenue_not_null,
    COUNT(shipping_revenue) AS shipping_revenue_not_null,
    COUNT(total_revenue) AS total_revenue_not_null
FROM gold.product_performance;
```

```sql
SELECT
    COUNT(*) AS total_rows,
    COUNT(product_category_name) AS category_not_null,
    COUNT(total_orders) AS total_orders_not_null,
    COUNT(total_items_sold) AS total_items_sold_not_null,
    COUNT(total_revenue) AS total_revenue_not_null,
    COUNT(average_product_price) AS average_price_not_null
FROM gold.category_performance;
```

```sql
SELECT
    COUNT(*) AS total_rows,
    COUNT(payment_type) AS payment_type_not_null,
    COUNT(total_orders) AS total_orders_not_null,
    COUNT(total_transactions) AS total_transactions_not_null,
    COUNT(total_payment_value) AS total_payment_value_not_null,
    COUNT(average_payment_value) AS average_payment_not_null
FROM gold.payment_performance;
```

✅Negative Values Check

```sql
SELECT *
FROM gold.customer_performance
WHERE total_orders < 0
   OR total_spent < 0
   OR average_order_value < 0;
```

```sql
SELECT *
FROM gold.product_performance
WHERE total_orders < 0
   OR product_revenue < 0
   OR shipping_revenue < 0
   OR total_revenue < 0;
```

```sql
SELECT *
FROM gold.category_performance
WHERE total_orders < 0
   OR total_items_sold < 0
   OR total_revenue < 0
   OR average_product_price < 0;
```

```sql
SELECT *
FROM gold.payment_performance
WHERE total_orders < 0
   OR total_transactions < 0
   OR total_payment_value < 0
   OR average_payment_value < 0;
```

✅Business Logic Checks

```sql
SELECT *
FROM gold.product_performance
WHERE ABS(
    (product_revenue + shipping_revenue) - total_revenue
) > 0.01;
```
