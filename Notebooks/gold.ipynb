#Daily_Sales_Table

```sql
SELECT
    o.order_id,
    o.order_purchase_timestamp_new,
    oi.price,
    oi.shipping_charges
FROM workspace.silver.orders o
JOIN workspace.silver.order_items oi
    ON o.order_id = oi.order_id
LIMIT 10;
```

```sql
CREATE OR REPLACE TABLE workspace.gold.daily_sales AS

SELECT
    CAST(o.order_purchase_timestamp_new AS DATE) AS order_date,

    COUNT(DISTINCT o.order_id) AS total_orders,

    SUM(oi.price) AS product_revenue,

    SUM(oi.shipping_charges) AS shipping_revenue,

    SUM(oi.price + oi.shipping_charges) AS total_revenue

FROM workspace.silver.orders o

JOIN workspace.silver.order_items oi
    ON o.order_id = oi.order_id

GROUP BY
    CAST(o.order_purchase_timestamp_new AS DATE);
```

```sql
SELECT *
FROM workspace.gold.daily_sales
ORDER BY order_date;
```

#product_Performance_Table

```sql
SELECT
    p.product_id,
    p.product_category_name,
    oi.order_id,
    oi.price,
    oi.shipping_charges
FROM workspace.silver.products p
JOIN workspace.silver.order_items oi
    ON p.product_id = oi.product_id
LIMIT 10;
```

```sql
CREATE OR REPLACE TABLE workspace.gold.product_performance AS

SELECT
    p.product_id,
    p.product_category_name,

    COUNT(DISTINCT oi.order_id) AS total_orders,

    ROUND(SUM(oi.price), 2) AS product_revenue,

    ROUND(SUM(oi.shipping_charges), 2) AS shipping_revenue,

    ROUND(SUM(oi.price + oi.shipping_charges), 2) AS total_revenue

FROM workspace.silver.products p

JOIN workspace.silver.order_items oi
    ON p.product_id = oi.product_id

GROUP BY
    p.product_id,
    p.product_category_name;
```

```sql
SELECT *
FROM workspace.gold.product_performance
ORDER BY total_revenue DESC
LIMIT 10;
```

#Customer_Performance_Table

```sql
SELECT
    o.customer_id,
    o.order_id,
    p.payment_value,
    p.payment_type
FROM silver.orders o
JOIN silver.payments p
    ON o.order_id = p.order_id
LIMIT 20;
```

```sql
SELECT
    o.customer_id,
    COUNT(DISTINCT o.order_id) AS total_orders,
    SUM(p.payment_value) AS total_spent,
    AVG(p.payment_value) AS average_order_value
FROM silver.orders o
JOIN silver.payments p
    ON o.order_id = p.order_id
GROUP BY o.customer_id
ORDER BY total_spent DESC
LIMIT 20;
```

```sql
CREATE OR REPLACE TABLE gold.customer_performance AS
SELECT
    o.customer_id,
    COUNT(DISTINCT o.order_id) AS total_orders,
    SUM(p.payment_value) AS total_spent,
    AVG(p.payment_value) AS average_order_value
FROM silver.orders o
JOIN silver.payments p
    ON o.order_id = p.order_id
GROUP BY o.customer_id;
```

```sql
SELECT *
FROM gold.customer_performance
ORDER BY total_spent DESC;
```

#Category_performance_Table

```sql
SELECT
    COALESCE(p.product_category_name, 'Unknown') AS product_category_name,
    COUNT(DISTINCT oi.order_id) AS total_orders,
    COUNT(oi.product_id) AS total_items_sold,
    SUM(oi.price) AS total_revenue,
    AVG(oi.price) AS average_product_price
FROM silver.order_items oi
JOIN silver.products p
    ON oi.product_id = p.product_id
GROUP BY COALESCE(p.product_category_name, 'Unknown')
ORDER BY total_revenue DESC;
```

```sql
CREATE OR REPLACE TABLE gold.category_performance AS

SELECT
    COALESCE(p.product_category_name, 'Unknown') AS product_category_name,
    COUNT(DISTINCT oi.order_id) AS total_orders,
    COUNT(oi.product_id) AS total_items_sold,
    ROUND(SUM(oi.price), 2) AS total_revenue,
    ROUND(AVG(oi.price), 2) AS average_product_price

FROM silver.order_items oi

JOIN silver.products p
    ON oi.product_id = p.product_id

GROUP BY COALESCE(p.product_category_name, 'Unknown');
```

```sql
SELECT *
FROM gold.category_performance
ORDER BY total_revenue DESC;
```

#Payment_Performance_Table

```sql
SELECT
    payment_type,
    COUNT(DISTINCT order_id) AS total_orders,
    COUNT(*) AS total_transactions,
    SUM(payment_value) AS total_payment_value,
    AVG(payment_value) AS average_payment_value
FROM silver.payments
GROUP BY payment_type
ORDER BY total_payment_value DESC;
```

```sql
CREATE OR REPLACE TABLE gold.payment_performance AS

SELECT
    payment_type,
    COUNT(DISTINCT order_id) AS total_orders,
    COUNT(*) AS total_transactions,
    ROUND(SUM(payment_value), 2) AS total_payment_value,
    ROUND(AVG(payment_value), 2) AS average_payment_value
FROM silver.payments
GROUP BY payment_type
ORDER BY total_payment_value DESC;
```

```sql
SELECT *
FROM gold.payment_performance;
```
